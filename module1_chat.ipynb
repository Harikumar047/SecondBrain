```python
import os
from dotenv import load_dotenv

load_dotenv()

if os.getenv("GROQ_API_KEY") is not None:
    print("GROQ API Key Found")
else:
    print("GROQ API Key Not Found")
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="openai/gpt-oss-120b", api_key=os.getenv("GROQ_API_KEY"), temperature=0.4)
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="openai/gpt-oss-120b", api_key=os.getenv("GROQ_API_KEY"), temperature=0.4)
```

```python
response = llm.invoke("Say hello in one sentence.")
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a personal productivity and knowledge assistant. Be concise."),
    ("human", "{input}")
])
```

```python
test_prompt = prompt_template.invoke({"input": "Summarize today's work."})
print(llm.invoke(test_prompt.messages).content)
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

memory = InMemorySaver()

agent = create_agent(
    llm,
    checkpointer=memory
)
```

```python
example_query = "My name is Hari."

events = agent.stream(
    {"messages": [("user", example_query)]},
    config={"configurable": {"thread_id": "a"}},
    stream_mode="values"
)

for event in events:
    event["messages"][-1].pretty_print()
```
